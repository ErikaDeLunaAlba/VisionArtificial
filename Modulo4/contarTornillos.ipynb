```python
import cv2
import numpy as np
```

```python

def contar_tornillos(ruta_imagen, area_min=100, area_max=10000, kernel_size=3, use_morph=True, use_equalize=False, gamma=None):
    imagen = cv2.imread(ruta_imagen, cv2.IMREAD_GRAYSCALE)
    if imagen is None:
        print(f"Error: No se pudo cargar la imagen {ruta_imagen}")
        return None
```

```python
# 2. Preprocesamiento por píxel (brillo/contraste/gamma/ecualización)
    if gamma is not None:
        # Corrección gamma
        imagen = np.power(imagen / 255.0, gamma) * 255.0
        imagen = np.clip(imagen, 0, 255).astype(np.uint8)

    if use_equalize:
        # Ecualización del histograma
        imagen = cv2.equalizeHist(imagen)
```

```python
 # 3. Binarización con Otsu (automática)
    # Nota: Usamos THRESH_BINARY_INV si los tornillos son oscuros sobre fondo claro; 
    # si son claros sobre fondo oscuro, usa THRESH_BINARY. 
    # En este ejemplo asumimos tornillos oscuros sobre fondo claro, así que usamos INV.
_, imagen_bin = cv2.threshold(imagen, 0, 255, cv2.THRESH_BINARY_INV | cv2.THRESH_OTSU)

    # 4. Operaciones morfológicas (opcional, especialmente para tornillos_2)
if use_morph:
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        # Apertura (erosión + dilatación) para eliminar pequeños puntos de ruido
        imagen_bin = cv2.morphologyEx(imagen_bin, cv2.MORPH_OPEN, kernel, iterations=1)
        # Opcional: Cierre para rellenar pequeños huecos dentro de los tornillos
        # imagen_bin = cv2.morphologyEx(imagen_bin, cv2.MORPH_CLOSE, kernel, iterations=1)
```

```python

    # 5. Etiquetado de componentes conexos
num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(imagen_bin, 8, cv2.CV_32S)
```

```python
# 6. Filtrado por área y conteo
    tornillos_detectados = 0
    # Crear una copia en color para dibujar los resultados
    imagen_color = cv2.cvtColor(imagen, cv2.COLOR_GRAY2BGR)

    for i in range(1, num_labels):  # i=0 es el fondo
        area = stats[i, cv2.CC_STAT_AREA]
        if area_min < area < area_max:
            tornillos_detectados += 1
            # Coordenadas del bounding box
            x = stats[i, cv2.CC_STAT_LEFT]
            y = stats[i, cv2.CC_STAT_TOP]
            w = stats[i, cv2.CC_STAT_WIDTH]
            h = stats[i, cv2.CC_STAT_HEIGHT]
            # Centroide
            cX, cY = centroids[i]
            # Dibujar rectángulo y centroide
            cv2.rectangle(imagen_color, (x, y), (x + w, y + h), (0, 255, 0), 2)
            cv2.circle(imagen_color, (int(cX), int(cY)), 4, (0, 0, 255), -1)
```

```python
 # Mostrar resultado
    print(f"Tornillos detectados en {ruta_imagen}: {tornillos_detectados}")
    cv2.imshow("Conteo de tornillos", imagen_color)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

    return tornillos_detectados

contar_tornillos("img/tornillos.jpg", area_min=50, area_max=5000, kernel_size=3, use_morph=True, use_equalize=False, gamma=1.2)
```
